```python

# Hidden Complexity Detector in Documents

import re

complex_words = {
    "utilize": "use",
    "commence": "start",
    "terminate": "end",
    "approximately": "about",
    "subsequently": "later",
    "endeavour": "try",
    "prior to": "before",
    "in the event of": "if",
    "at this point in time": "now",
    "in order to": "to",
    "sufficient": "enough",
    "assistance": "help"
}

def analyze_document(text):
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    flagged = []

    print("=" * 50)
    print("HIDDEN COMPLEXITY DETECTOR")
    print("=" * 50)

    for sentence in sentences:
        if not sentence:
            continue

        words = re.findall(r"\b[\w'-]+\b", sentence)
        found = [
            phrase for phrase in complex_words
            if re.search(
                r"\b" + re.escape(phrase) + r"\b",
                sentence,
                re.IGNORECASE
            )
        ]

        # Flag long sentences or sentences with complex phrases
        if len(words) > 15 or found:
            flagged.append(sentence)
            print("\nComplex sentence:", sentence)
            print("Word count:", len(words))

            if found:
                print("Difficult phrases:", ", ".join(found))

                simpler = sentence
                for phrase in sorted(
                    found, key=len, reverse=True
                ):
                    simpler = re.sub(
                        r"\b" + re.escape(phrase) + r"\b",
                        complex_words[phrase],
                        simpler,
                        flags=re.IGNORECASE
                    )

                print("Simpler version:", simpler)
            else:
                print("Suggestion: Split this long sentence into shorter sentences.")

    print("\nTotal sentences:", len([s for s in sentences if s]))
    print("Flagged sentences:", len(flagged))
    print("=" * 50)


# Sample input
document = (
    "Students must utilize the online portal in order to submit their forms. "
    "The examination will commence on Monday. "
    "Bring your identity card."
)

analyze_document(document)
```
